```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import io, segmentation, color, filters, feature, measure
```

```python
image = io.imread('aLDNHwu.png')
```

```python
image = color.rgb2gray(image)
```

```python
image = filters.gaussian(image, sigma=1.5)
```

```python
edges = feature.canny(image, sigma=2.5)
```

```python
segments = segmentation.watershed(edges, markers=np.zeros_like(image, dtype=int))
```

```python
labels = measure.label(segments)
```

```python
plt.figure(figsize=(10, 10))
plt.imshow(color.label2rgb(labels, image=edges))
plt.axis('on')
plt.show()
```

```python
io.imsave('output.png', labels)
```
