```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
import os
import cv2
import numpy as np
from keras.preprocessing.image import img_to_array, load_img

def load_data(data_path):
    images = []
    labels = []

    for filename in os.listdir(data_path + 'train-volume/'):
        img = load_img(os.path.join(data_path + 'train-volume', filename), color_mode='grayscale')
        img_array = img_to_array(img)
        images.append(img_array)
    
    for filename_label in os.listdir(data_path + 'train-labels/'):
        label = load_img(os.path.join(data_path + 'train-labels/', filename_label), color_mode='grayscale')
        label_array = img_to_array(label)
        labels.append(label_array)

    return np.array(images), np.array(labels)

data_path = 'dataset/'
train_images, train_labels = load_data(data_path)

# Normalize images
train_images = train_images / 255.0
train_labels = train_labels / 255.0
```

```python
from keras.models import Model
from keras.layers import Input, Conv2D, MaxPooling2D, UpSampling2D

def build_model():
    input_layer = Input(shape=(512, 512, 1))

    # Encoder
    conv1 = Conv2D(32, (3, 3), activation='relu', padding='same')(input_layer)
    pool1 = MaxPooling2D(pool_size=(2, 2))(conv1)

    # Decoder
    up1 = UpSampling2D(size=(2, 2))(pool1)
    decoded = Conv2D(1, (3, 3), activation='sigmoid', padding='same')(up1)

    model = Model(inputs=input_layer, outputs=decoded)
    model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

    return model

model = build_model()
```

```python
from sklearn.model_selection import train_test_split

# Assuming train_images and train_labels are already loaded and preprocessed
X_train, X_val, y_train, y_val = train_test_split(train_images, train_labels, test_size=0.2, random_state=42)

model.fit(X_train, y_train, validation_data=(X_val, y_val), epochs=10, batch_size=32)
model.save('Edge1.h5')
```

```python
# Load and preprocess your test set
test_images, test_labels = load_data('dataset/')
test_images = test_images / 255.0
test_labels = test_labels / 255.0

# Evaluate the model
eval_result = model.evaluate(test_images, test_labels)
print(f"Test Loss: {eval_result[0]}, Test Accuracy: {eval_result[1]}")
```

```python
import matplotlib.pyplot as plt

# Assuming you have a new image for prediction
new_image_path = 'train-volume/train-volume_19.tif'
new_image = load_img(new_image_path, color_mode='grayscale')
new_image_array = img_to_array(new_image)
new_image_array = new_image_array / 255.0

# Reshape the image to match the model input shape
new_image_array = np.reshape(new_image_array, (1, 512, 512, 1))

# Make predictions
predictions = model.predict(new_image_array)

# Apply threshold to create a binary image
threshold = 0.5798  # Adjust this threshold as needed
binary_predictions = (predictions > threshold).astype(np.uint8)

# Display the original image
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.imshow(new_image_array[0, :, :, 0], cmap='gray')  # Assuming the image is 2D (height, width, channels)
plt.title('Original Image')

# Display the predicted edges with black border and white background
plt.subplot(1, 2, 2)
plt.imshow(binary_predictions[0, :, :, 0], cmap='gray', vmin=0, vmax=1)
plt.title('Predicted Edges')

plt.show()
```
